```python
import numpy as np

math = np.array([88, 76, 95, 65, 82, 91, 70, 88, 55, 92])

print(math)
```

```python
print("Mean:", np.mean(math))
print("Median:", np.median(math))
print("Standard deviation:", np.std(math))
print("Maximum:", np.max(math))
print("Minimum:", np.min(math))
```

We can put Maths, Science and English marks into one array.

Each row-> one student
each coloumn-> subjects (Maths, English, Science)

```python
marks = np.array([
    [88, 92, 85],
    [76, 81, 90],
    [95, 89, 88],
    [65, 70, 78],
    [82, 85, 80],
    [91, 94, 92],
    [70, 68, 72],
    [88, 82, 75],
    [55, 60, 68],
    [92, 95, 90]
])
```

axis = 1 -> each student/row

```python
np.sum(marks, axis=1)
```

axis = 0 -> each subject/coloumn

```python
np.mean(marks, axis=0)
```

np.where() allows us to make a decision based on a condition.

```python
print(len(np.where(math > 80)[0]))
```

Ann had 55 marks in maths
and the array before [88 76 95 65 82 91 70 88 55 92]

```python
new_math = np.where(math < 60, math + 5, math)

print(new_math)
```

Do more study hours relate to higher Maths marks?

0 -> first row (study hours)
1 -> second column (maths)

```python
study_hours = np.array([14,8,18,5,10,16,6,7,2,8])
```

```python
correlation = np.corrcoef(study_hours, math)[0, 1]

print(correlation)
```
