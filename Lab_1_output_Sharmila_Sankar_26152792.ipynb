## 2. Input and arithmetic

Ask the user for the number of records before and after cleaning. Convert both inputs to integers. Print the number of records removed and the percentage retained, rounded to one decimal place. Assume the initial count is greater than zero.


```python
# Write your code here
initial_records = int(input("Enter number of records before cleaning: "))
cleaned_records = int(input("Enter number of records after cleaning: "))

records_removed = initial_records - cleaned_records
percentage_retained = (cleaned_records / initial_records) * 100

print(f"Records removed: {records_removed}")
print(f"Percentage retained: {round(percentage_retained, 1)}%")
```

## 7. Functions

Write a function `average_score(scores)` that returns the average of a non-empty list of scores. Call it with `[70, 80, 90]` and print the result.


```python
# Write your code here
def average_score(scores):
  return sum(scores) / len(scores)

result = average_score([70, 80, 90])

print(f"Average score: {result}")
```

## 11. Built-in functions

Given `values = [-3, 8, 0, -5, 10]`, print the minimum, maximum, sum and number of values. Then use a loop and `abs()` to print each absolute value.


```python
# Write your code here
values = [-3, 8, 0, -5, 10]

minimum = min(values)
maximum = max(values)
total = sum(values)
count = len(values)

print(f"Minimum: {minimum}")
print(f"Maximum: {maximum}")
print(f"Sum: {total}")
print(f"Number of values: {count}")

for value in values:
  print(abs(value))
```

## 12. Nested data challenge

Given the records below, use a loop to print the name of each student whose score is at least 50. Then calculate and print the mean score. Use only concepts covered in this lab.

`students = [{"name": "Amina", "score": 72}, {"name": "Ben", "score": 45}, {"name": "Chen", "score": 83}]`


```python
# Write your code here
students = [{"name": "Amina", "score": 72}, {"name": "Ben", "score": 45}, {"name": "Chen", "score": 83}]

for student in students:
  if student["score"] >= 50:
    print(f"{student['name']}")

#Calculate the mean score
total = 0

for student in students:
  total = total + student["score"]

mean_score = total / len(students)

print(f"Mean score:{(mean_score)}")
```
